```python
import os
import glob
import pandas as pd

# 1. Buscamos automáticamente los archivos en las carpetas más comunes
def buscar_archivo(nombre_parcial):
    # Carpetas típicas del sistema donde suelen estar las descargas o proyectos
    carpetas_busqueda = [
        '.',                          # Carpeta actual del notebook
        os.path.expanduser('~/Downloads'),  # Descargas (Mac/Linux/Windows)
        os.path.expanduser('~/Desktop'),    # Escritorio
        os.path.expanduser('~/Documents')   # Documentos
    ]
    
    for carpeta in carpetas_busqueda:
        # Buscamos coincidencias ignorando mayúsculas/minúsculas y variaciones de extensión
        patron = os.path.join(carpeta, f"*{nombre_parcial}*")
        archivos_encontrados = glob.glob(patron)
        if archivos_encontrados:
            print(f"✅ Encontrado '{nombre_parcial}' en: {archivos_encontrados[0]}")
            return archivos_encontrados[0]
            
    raise FileNotFoundError(f"❌ No se encontró ningún archivo que contenga '{nombre_parcial}' en Descargas, Escritorio o la carpeta actual.")

try:
    # 2. Localizar las rutas reales de tus archivos (Cambiado 'kaggle' por 'reporte_matriz_experimentos')
    ruta_li01 = buscar_archivo('li01_nhc')
    ruta_li02 = buscar_archivo('li02_nhc')
    ruta_matriz = buscar_archivo('reporte_matriz_experimentos')

    print("\n🔄 Procesando y uniendo los datos...")
    
    # 3. Cargar los archivos CSV detectados
    df_li01 = pd.read_csv(ruta_li01)
    df_li02 = pd.read_csv(ruta_li02)
    df_matriz = pd.read_csv(ruta_matriz)

    # 4. Concatenar de forma vertical los experimentos
    df_experimentos_todos = pd.concat([df_li01, df_li02], ignore_index=True)

    # 5. CORRECCIÓN CLAVE: Agregar el prefijo "KA" si empiezan con "z" para homologar con la matriz
    df_experimentos_todos['experimento'] = df_experimentos_todos['experimento'].apply(
        lambda x: f"KA{x}" if str(x).startswith('z') else x
    )

    # 6. Relacionar por la columna "experimento" con el archivo reporte_matriz_experimentos
    df_resultado = pd.merge(df_experimentos_todos, df_matriz, on='experimento', how='inner')

    # 7. Guardar el resultado en la carpeta actual del notebook
    df_resultado.to_csv('_experimentos_p_k_c_matriz.csv', index=False)
    print("💾 ¡Archivo '_experimentos_p_k_c_matriz.csv' exportado con éxito!")

    # 8. Mostrar la vista previa de la tabla en Jupyter
    df_resultado

except FileNotFoundError as e:
    print(e)
    print("\n👉 Solución manual alternativa: Mueve o arrastra físicamente tus 3 archivos CSV a la misma carpeta donde tienes guardado este notebook de Jupyter.")
```
